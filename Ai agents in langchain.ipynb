```python
from langchain_ollama import OllamaLLM
import pandas as pd
```

```python
from langchain_core.tools import tool
import requests
from langchain_community.tools import DuckDuckGoSearchRun
```

```python
search_tool = DuckDuckGoSearchRun()
```

```python
search_tool.invoke('top 3 news in india today')
```

```python
llm = OllamaLLM(model='qwen3:8b')
```

```python
llm.invoke('Who are you?')
```

```python
from langchain.agents import create_react_agent,AgentExecutor
from langchain import hub
```

```python
prompt = hub.pull('hwchase17/react')
```

```python
agent = create_react_agent(tools = [search_tool],llm = llm,prompt = prompt)
```

```python
executer = AgentExecutor(agent=agent,tools=[search_tool],verbose=True,handle_parsing_errors=True)
```

```python
executer.invoke({'input':'analyse the latest news and give me the 3 emerging sectors for investment'})
```
