```python
from langgraph.graph import StateGraph,START,END
import operator
from typing import TypedDict,Annotated
# from langchain_core import JsonOutputParser
from langchain_ollama import OllamaLLM
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field
```

```python
class EvaluationState(TypedDict):
    article:str
    language_summary: str
    technical_depth_summary: str
    clarity_of_though: str
    overall: str
    score: Annotated[list[int],operator.add]
    avg_score: float
```

```python
class EvaluationSchema(BaseModel):
    summary: str = Field(Description = "Summary of the Article in 50 words")
    score: int = Field(Description = "Score out of 10",ge=0,le=10)
```

```python
model = OllamaLLM(model = "gemma2:2b")
```

```python
parser = JsonOutputParser(pydantic_object=EvaluationSchema)
```

```python
def evaluate_language(state: EvaluationState):
    prompt = f"""Return ONLY valid JSON with this structure:
{{"summary": "...", "score": }}

Evaluate the language quality of the essay and provide feedback and assign a score out of 10 \n {state['article']}"""
    chain = model | parser
    result = chain.invoke(prompt)
    return {'language_summary':result['summary'],'score':[result['score']]}
```

```python
def evaluate_technical(state: EvaluationState):
    prompt = f"""Return ONLY valid JSON with this structure:
{{"summary": "...", "score": }}

Evaluate the technical depth of the essay and provide feedback and assign a score out of 10 \n {state['article']}"""
    chain = model | parser
    result = chain.invoke(prompt)
    return {'technical_depth_summary':result['summary'],'score':[result['score']]}
```

```python
def evaluate_clarity(state: EvaluationState):
    prompt = f"""Return ONLY valid JSON with this structure:
{{"summary": "...", "score": }}

Evaluate the clarity of thought of the essay and provide feedback and assign a score out of 10 \n {state['article']}"""
    chain = model | parser
    result = chain.invoke(prompt)
    return {'clarity_of_though':result['summary'],'score':[result['score']]}
```

```python
def final_evaluation(state: EvaluationState):
    prompt = f"""Return ONLY valid JSON with this structure:
{{"summary": "...", "score": }}

Summarise overall essay \n {state['article']}"""
    chain = model | parser
    result = chain.invoke(prompt)

    avg_score = sum(state['score'])/len(state['score']) 

    return {'overall':result['summary'],'avg_score':avg_score}
```

```python
graph = StateGraph(EvaluationState)
graph.add_node('evaluate_language',evaluate_language)
graph.add_node('evaluate_technical',evaluate_technical)
graph.add_node('evaluate_clarity',evaluate_clarity)
graph.add_node('final_evaluation',final_evaluation)
```

```python
graph.add_edge(START,'evaluate_language')
graph.add_edge(START,'evaluate_technical')
graph.add_edge(START,'evaluate_clarity')

graph.add_edge('evaluate_language','final_evaluation')
graph.add_edge('evaluate_technical','final_evaluation')
graph.add_edge('evaluate_clarity','final_evaluation')

graph.add_edge('final_evaluation',END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {'article':article}
```

```python
workflow.invoke(initial_state)
```
